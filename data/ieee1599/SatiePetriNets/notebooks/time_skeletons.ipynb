## Authoritative Timeline

```python
from timetoalign.loader.score import Ms3Loader
from timetoalign.timelines import SegmentLine
from timetoalign.timelines import ContinuousLogicalTimeline

ms3_measures = "trois-gymnopedies.measures.tsv"
ms3_loader = Ms3Loader.from_file(ms3_measures)
gymn_timeline = ms3_loader.create_timeline()
gymn_timeline
```

```python
flow_controller = ms3_loader.create_flow_controller()
flow_controller
```

```python
boundaries = [0] + [
    (section.mc_end - 1) * 3 # 3 quarters per measure
    for section in flow_controller.iter_sections()
]
gymn_timeline.create_regions_from_boundaries(boundaries, prefix="gymno")
gymn_timeline.create_children_from_regions()
gymn_timeline
```

## Orchestra timeline

```python
orch_timeline = SegmentLine(ContinuousLogicalTimeline, unit="q")
for piece in (3, 1, 2):
    orch_timeline.append_segment(
        gymn_timeline.get_child(f"gymno_{piece}")
    )
orch_timeline
```
